```python
## Applying a simple transformation -> convert the text to upper case

from datasets import Dataset


custom_data = {
    "text": ["I love Transformers!", "Hugging Face makes NLP easy.", "Datasets library is amazing!"],
    "label": [1, 1, 1]
}

dataset = Dataset.from_dict(custom_data)

# Apply transformation: Convert all text to uppercase
def uppercase_text(example):
  ## Each row is an example
  example["text"] = example["text"].upper()
  return example

# Apply transformation to the dataset
transformed_dataset = dataset.map(uppercase_text)

print(transformed_dataset)
```

## Tokenizer

Reference: https://huggingface.co/docs/transformers/preprocessing

Once we are good with the data and all the transformation on the Natural Language side, it's time to prepare these to be converted to vectors. The tokenizer converts raw text to format that can be consumed by models.

A tokenizer is responsible for preprocessing text into an array of numbers as inputs to a model. There are multiple rules that govern the tokenization process, including how to split a word and at what level words should be split. The most important thing to remember is you need to instantiate a tokenizer with the same model name to ensure you're using the same tokenization rules a model was pretrained with.

### Loading and using a pretrained tokenizer

```python
## Getting started with a pre-trained tokenizer

from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("google-bert/bert-base-cased")
```

```python
from pprint import pprint
encoded_input = tokenizer("Natural Language processing classes are on every wednesday !! It's fun and amazing")
pprint(encoded_input)
```

The tokenizer returns 3 things:
- input_ids: This is the index corresponding to each token in the sentence
- attention_mask: indicates whether a token should be attended or not
- token_type_ids: identifies which sequence a token belongs to when there is more than one sequence

```python
## Getting back the text using the token ids

tokenizer.decode(encoded_input["input_ids"])
```

The tokenizer adds 2 special tokens -- CLS & SEP (Classifier and Seperator) to the sentence

### Batched Sentences

In case of multiple sentence we can pass them as a list to tokenizer

```python
batch_sentences = [
    "But what about second breakfast?",
    "Don't think he knows about second breakfast, Pip.",
    "What about elevensies?",
]
encoded_inputs = tokenizer(batch_sentences)
print(encoded_inputs)
```

```python
for ele in encoded_inputs["input_ids"]:
  print(f"Length of sentence is : {len(ele)}")
```

### Padding

The sentence are generally of different lengths and this can be an issue because the tensors which are model inputs need to be of uniform shape, hence we perform padding by adding a special padding token to shorter sentences

```python
batch_sentences = [
    "But what about second breakfast?",
    "Don't think he knows about second breakfast, Pip.",
    "What about elevensies?",
]
encoded_inputs = tokenizer(batch_sentences, padding=True) ## Pads the shorter sentence to the lnegth of the longest sentence
print(encoded_inputs)
```

```python
for ele in encoded_inputs["input_ids"]:

  print(f"Length of sentence is : {len(ele)}")
```

As you can see the length of each sentence is same after padding

```python
## Decoding these sentences gives us the [PAD] token
print(tokenizer.decode(encoded_inputs["input_ids"][0]))
print(tokenizer.decode(encoded_inputs["input_ids"][1]))
print(tokenizer.decode(encoded_inputs["input_ids"][2]))
```

### Truncation

In some cases the model will have limitation on the length of sentence for performance reasons. For example for BERT model, the maximum length is 512 tokens. The mode might nor process longer sentences due to memory and computation constraints. Due to these reasons we want to truncate the sentence to a fixed number of tokens.

```python
batch_sentences = [
    "I was literally starving by that time since I had not eating anything that morning.",
    "It was such a good day.",
    "How many apples should I buy Mark?",
]
encoded_inputs = tokenizer(batch_sentences, truncation=True, max_length = 5) ## truncated to 5 tokens
non_truncated_inputs = tokenizer(batch_sentences)
print(encoded_inputs)
```

```python
for ele in non_truncated_inputs["input_ids"]:
  print(f"Length of non truncated sentence is : {len(ele)}")

for ele in encoded_inputs["input_ids"]:
  print(f"Length of truncated sentence is : {len(ele)}")
```

The tokenizer config related to model will have the parameter max_length which truncates the sentence in case truncation is set to True

## Fine-Tuning a pre-trained model

Reference: https://huggingface.co/docs/transformers/training

### Training Preparation

```python
## First let's load the dataset

from datasets import load_dataset

# Load AG News dataset
train_dataset = load_dataset("ag_news", split="train[:500]") ## Let's take 500 examples for training

# For test we will use the same earlier samples
```

```python
train_dataset[0]
```

```python
## Define tokenizer
from transformers import AutoTokenizer

model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
## Tokenize the data

def tokenize_example(example):
  tokenized_example = tokenizer(example["text"], max_length=128, padding="max_length", truncation=True)
  tokenized_example["label"] = example["label"]
  return tokenized_example

train_dataset = train_dataset.map(tokenize_example, remove_columns=["text"]) ## We don't need text column after tokenization in the dataset
```

```python
## Apply transformation to the test

test_dataset = dataset.map(tokenize_example, remove_columns=["text"])
```

```python
print(train_dataset[0])
```

```python
## Prepare for training
train_dataset = train_dataset.rename_column("label", "labels")
train_dataset.set_format("torch")

test_dataset = test_dataset.rename_column("label", "labels")
test_dataset.set_format("torch")
```

```python
## Define a DataLoader -- This is a torch utility that helps create iterations of data that can be given to the model
from torch.utils.data import DataLoader

train_dataloader = DataLoader(train_dataset, shuffle=True, batch_size=8)
test_dataloader = DataLoader(test_dataset, batch_size=2)
```

### Using PyTorch Training Loop

```python
## Define Model
from transformers import AutoModelForSequenceClassification
from torch.optim import AdamW
from transformers import get_scheduler

model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=4)
device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)

## Definfe optimizer
optimizer = AdamW(model.parameters(), lr=5e-5)

## Define Loss Function
loss_fn = torch.nn.CrossEntropyLoss()

## Learning rate scheduler
num_epochs = 3
num_training_steps = num_epochs * len(train_dataloader)
lr_scheduler = get_scheduler(
    name="linear", optimizer=optimizer, num_warmup_steps=0, num_training_steps=num_training_steps
)
```

```python
for batch in train_dataloader:
  print(batch.keys())
  print(batch["labels"])
  break
```

```python
## Training Loop
from tqdm.notebook import tqdm
progress_bar = tqdm(range(num_training_steps))
model.train()
for epoch in range(num_epochs):
    for batch in train_dataloader:
        labels = batch["labels"].to(device)
        input_ids = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)

        outputs = model(input_ids=input_ids, attention_mask=attention_mask)

        loss = loss_fn(outputs.logits, labels)
        loss.backward()

        optimizer.step()
        lr_scheduler.step()
        optimizer.zero_grad()
        progress_bar.update(1)
```

```python
## Testing
import evaluate
model.eval()
metric = evaluate.load("accuracy")
for batch in test_dataloader:
  input_ids = batch["input_ids"].to(device)
  attention_mask = batch["attention_mask"].to(device)
  with torch.no_grad():
    output = model(input_ids=input_ids, attention_mask=attention_mask)

  logits = output.logits.detach().cpu()
  prediction = torch.argmax(logits, dim=-1)
  metric.add_batch(predictions=prediction, references=batch["labels"])

print("Accuracy: ", metric.compute())
```

With Fine tuning we were able to increase the accuracy from 18% to 78%

### Using The Huggingface Trainer

```python
from transformers import TrainingArguments, Trainer
import evaluate
import numpy as np

model = AutoModelForSequenceClassification.from_pretrained("bert-base-uncased", num_labels=4)

## Contains all the hyperparameters and arguments for training
training_args = TrainingArguments(output_dir="test_trainer", eval_strategy="epoch", report_to="none")

## Define the function to use for evaluation
## All models return logits which we need to process to compute our metric (accuracy in this case)
metric = evaluate.load("accuracy")
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    return metric.compute(predictions=predictions, references=labels)

# Define the trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    compute_metrics=compute_metrics,
)

trainer.train()
```

## Parameter Efficient Fine Tuning & Quantization

Resources: \\
https://huggingface.co/docs/transformers/peft \\
https://huggingface.co/docs/peft/index

As the langauge models become bigger in size, they encapsulate the ability to perform wide range of tasks of different domains. Now if we want to focus on specific domain or a specific task, we wouldn't need to fine tune the whole of Language Models. Here is where PEFT or Parameter Efficient Fine tuning comes into play. It helps use fine-tune some very speicifc weigths so that we can fine tune the langauge model for our specific need and make it "adapt" to it.

Also, as the models grow bigger in size, loading the model weights in memory causes memory constraint. Hence we perform quantization in some case where we load the model in a 4-bit or 8-bit kind of setting.

For this exercise let's look at LoRA, a popular PEFT method and some Quantization methods.

### PEFT

```python
!pip install peft
```

LoRA or Low Rank Adapters is a method where the model weights are frozen and some additional weights are trained and then added to the model weights. The term "Low Rank" comes from the fact that, we want to project a matix of the similar shape as our Weight Matrices to a low rank projection as a product of 2 matrices using the rank and then train the params of these 2 matrices.

If thr size of matrix W is NxN, we want to break it as W = AB, where A is Nxr and B is rxN and the number of parameters reduce from NxN to 2Nr (r<<N)

```python
## Let's load a model and add adapters to it

from transformers import AutoModelForCausalLM, OPTForCausalLM, AutoTokenizer
from peft import LoraConfig

model_id = "facebook/opt-350m"
model = AutoModelForCausalLM.from_pretrained(model_id)

## We have to define a LoRA config, the target modules are basically the weight matrices where we add adapter weights after training
lora_config = LoraConfig(
    target_modules=["q_proj", "k_proj"],
    init_lora_weights=False,
    r=8 ## Rank
)

model.add_adapter(lora_config, adapter_name="adapter_1")
```

```python
## Print Model parameters

print(f"Number of Parameters: {model.num_parameters()}")

## Print Trainable parameters

print(f"Number of Trainable Parameters: {model.num_parameters(only_trainable=True)}")
```

As you can see the number of parameters have greatly reduced

```python
## Using a pretreained-adapter
from peft import PeftConfig

model_id = "facebook/opt-350m"
adapter_model_id = "ybelkada/opt-350m-lora"

model = AutoModelForCausalLM.from_pretrained(model_id)
peft_config = PeftConfig.from_pretrained(adapter_model_id) ## Load the adapter config
```

```python
input = "Hello "
tokenizer = AutoTokenizer.from_pretrained(model_id)
input_ids = tokenizer(input, return_tensors="pt").input_ids

model.add_adapter(peft_config) ## Add adapter with the pretrained adapter config
model.enable_adapters() ## Enable Adapters

output = model.generate(input_ids)
print(tokenizer.decode(output[0], skip_special_tokens=True))
```

```python
## We can also disabile the adapters
model.disable_adapters()
output = model.generate(input_ids)
print(tokenizer.decode(output[0], skip_special_tokens=True))
```

### Quantization

Quantization techniques focus on representing data with less information while also trying to not lose too much accuracy. This often means converting a data type to represent the same information with fewer bits. For example, if your model weights are stored as 32-bit floating points and they're quantized to 16-bit floating points, this halves the model size which makes it easier to store and reduces memory-usage.

```python
pip install -U bitsandbytes
```

```python
from transformers import AutoModelForSequenceClassification
import torch
import os

model_name = "bert-base-uncased"

model_fp32 = AutoModelForSequenceClassification.from_pretrained(model_name) ## We are Loading the model in unquantized form

fp32_path = "model_fp32"
model_fp32.save_pretrained(fp32_path)

## Calculate the size of the non-quantized model
def get_size(model_path):
    total_size = sum(os.path.getsize(os.path.join(model_path, f)) for f in os.listdir(model_path)) / (1024 * 1024)
    return total_size

size_fp32 = get_size(fp32_path)
print(f"Model size (FP32): {size_fp32:.2f} MB") ## This is the unquantized model

model_int8 = AutoModelForSequenceClassification.from_pretrained(model_name, load_in_8bit=True) ## Loading the model in 8bit (Quantized)

int8_path = "model_int8"
model_int8.save_pretrained(int8_path)


size_int8 = get_size(int8_path)
print(f"Model size (INT8): {size_int8:.2f} MB") ## This is the quantized model
```

You can see the size of model reduces when loaded in quantized form

## Text Generation

Reference: https://huggingface.co/docs/transformers/llm_tutorial

Now we will look at Text Generation. In the above examples we looked at different tasks like sentiment analysis, text classification, No we will look at Text Generation.

Large Language Models act as a key component for text generation, they are large pretrained transformer models trained to predict next word.

These models do something known as Auto-Regressive decoding, which is predicting one word/token at a time. A key aspect of this is that the model provides a probabilistic distribution over all the words in the vocab and then based on some strategies the next word is selected.

### Basics

```python
## We will look into the gpt-2 model
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = AutoModelForCausalLM.from_pretrained("gpt2", pad_token_id=tokenizer.eos_token_id)

input_text = "Today is a shiny day and hence I am "
input_ids = tokenizer(input_text, return_tensors="pt").input_ids

greedy_output = model.generate(input_ids, max_length=50) ## Here we will select the token that has the maximum probability
print(tokenizer.decode(greedy_output[0], skip_special_tokens=True))
```

```python
print(tokenizer.decode(greedy_output[0], skip_special_tokens=True))
```

Now the key idea in these models is that they are stochastic and the stochasticity is goverened by the probability distribution over the words. We can use different decoding methods other than the greedy decoding. The options are: Beam Search, Sampling

```python
## Beam Seach decoding
beam_output = model.generate(
    input_ids,
    max_length=50,
    num_beams=5,
    early_stopping=True
)
print(f"Answer: ")
print(tokenizer.decode(beam_output[0], skip_special_tokens=True))
```

```python
## Sampling -- Randomg sampling based on the probability distribution

sampling_output = model.generate(
    input_ids,
    max_length=50,
    do_sample=True,
)

print(f"Answer: ")
print(tokenizer.decode(sampling_output[0], skip_special_tokens=True))
```

```python
## You can control the sampling using different params -- Temperature, Top-K, Top-P
## You will learn more about these in the upcoming sections

topk_output = model.generate(
    input_ids,
    max_length=50,
    do_sample=True,
    top_k=10
)

print(f"-"*50)
print(f" Top-K setting Answer: ")
print(tokenizer.decode(topk_output[0], skip_special_tokens=True))
print("\n\n")

topp_output = model.generate(
    input_ids,
    max_length=50,
    do_sample=True,
    top_p=0.5
)

print(f"-"*50)
print(f"Top-P setting Answer: ")
print(tokenizer.decode(topp_output[0], skip_special_tokens=True))
print("\n\n")

temp_output = model.generate(
    input_ids,
    max_length=50,
    do_sample=True,
    temperature=1.5
)
print(f"-"*50)
print(f"Temperature Setting Answer: ")
print(tokenizer.decode(temp_output[0], skip_special_tokens=True))
print("\n\n")
```
